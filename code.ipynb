####  Two sample t test on Park faci column

```python
df[df.PARK_FACIL == 'Yes']['SALES_PRICE'].values
```

```python
from scipy.stats import ttest_ind

# Class A and B scores
class_a = df[df.PARK_FACIL == 'Yes']['SALES_PRICE'].values
class_b = df[df.PARK_FACIL == 'No']['SALES_PRICE'].values

# Perform two-sample t-test (assume equal variances)
t_stat, p_val = ttest_ind(class_a, class_b)

print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.6f}")

# Interpretation
alpha = 0.05
if p_val < alpha:
    print("Reject the null hypothesis: The class averages are significantly different.")
else:
    print("Fail to reject the null hypothesis: No significant difference.")
```

## Prep Data for ML

```python
df = df[['AREA','BUILDING_AGE', 'INT_SQFT', 'DATE_SALE', 'DIST_MAINROAD', 'N_BEDROOM',
       'N_BATHROOM', 'N_ROOM', 'SALE_COND', 'PARK_FACIL', 'DATE_BUILD',
       'BUILDTYPE', 'UTILITY_AVAIL', 'STREET', 'MZZONE', 'QS_ROOMS',
       'QS_BATHROOM', 'QS_BEDROOM', 'QS_OVERALL', 'SALES_PRICE']]
```

```python
# Filter out categorical column
X_cat = df.select_dtypes('object')
X_cat
```

```python
order_col = [list(df.groupby(i)['SALES_PRICE'].mean().sort_values(ascending=False).index) for i in X_cat.columns ]
```

```python
# encode the columns
from sklearn.preprocessing import OrdinalEncoder
order = OrdinalEncoder(categories=order_col,dtype='int64')
X_cat_En = pd.DataFrame(order.fit_transform(X_cat),columns=order.get_feature_names_out())
X_cat_En
```

```python
# separate 
X_cat_En1 = df[['N_BEDROOM','N_BATHROOM','N_ROOM']]
X_cat_En1.reset_index(drop=True,inplace=True)
```

```python
y = df['SALES_PRICE']
```

Numerical columns encoding using standardization

```python
X_num = df.select_dtypes(['int32','int64','float'])
X_num.drop(columns=['N_BEDROOM','N_BATHROOM','N_ROOM'],inplace=True)
X_num.reset_index(drop=True,inplace=True)
```

```python
X_num.corr()['SALES_PRICE'].sort_values(ascending=False)
```

```python
X_num = X_num[['INT_SQFT','BUILDING_AGE']]
```

```python
from sklearn.preprocessing import StandardScaler
Scale = StandardScaler()
X_num_en = pd.DataFrame(Scale.fit_transform(X_num),columns=Scale.get_feature_names_out())
X_num_en.reset_index(drop=True,inplace=True)
```

```python
df
```

```python
x = pd.concat([X_cat_En,X_cat_En1,X_num_en],axis=1)
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

### Model preparation

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score,mean_absolute_error

models = {
    'LR' : LinearRegression(),
    'svr' : SVR(),
    'DT' : DecisionTreeRegressor(),
    'RFR' : RandomForestRegressor()
}
```

```python
for name,model in models.items():
    model.fit(x_train,y_train)

    y_train_pread = model.predict(x_train)
    y_test_pread = model.predict(x_test)

    print(15*'-',model,15*'-')
    print(' ')
    print('Train data Prediction')
    print('R2 Score :',r2_score(y_train,y_train_pread))
    print('MAE :',mean_absolute_error(y_train,y_train_pread))
    print('Test data Prediction')
    print('R2 Score :',r2_score(y_test,y_test_pread))
    print('MAE :',mean_absolute_error(y_test,y_test_pread))
    print(' ')
    print(15*'*',model,15*'*')
    print(' ')
    print(' ')
```

# 🔹 Linear Regression

    1.Train R²: 0.88 | Test R²: 0.886 → almost equal → no overfitting.

    2.MAE ~10 lakhs → relatively high error compared to tree-based models.

✅ Good, simple baseline.

# 🔹 SVR (Support Vector Regression)

    1.Train/Test R² ~ -0.02 → SVR is failing to learn patterns.

    2.MAE ~29 lakhs → much worse than others.

❌ Reason:

    SVR is very sensitive to feature scaling (needs StandardScaler or MinMaxScaler).

    Default C=1.0, epsilon=0.1 aren’t suitable for large target values.

    That’s why it looks almost like predicting the mean of y.

# 🔹 Decision Tree Regressor

    1.Train R² = 1.0, MAE = 0 → overfitting (memorized training data).

    2.Test R² = 0.973 → still strong performance.

✅ Can be improved by setting max_depth, min_samples_split, etc., to reduce overfit.

# 🔹 Random Forest Regressor

    1.Train R² = 0.998 | Test R² = 0.989 → near-perfect but with good generalization.

    2.MAE ~3 lakhs → far better than Linear Regression.

✅ This is your best model.

# ✅ Takeaways

1.Best model: RandomForestRegressor.

2.Stable baseline: LinearRegression.

3.Overfit but decent: DecisionTreeRegressor.

4.Failed model: SVR (needs scaling + tuning).
